# **Creating Target Variable**







```python
def get_result(row):
  if row["home_score"] > row["away_score"]:
    return "Home_win"
  elif row["home_score"] < row["away_score"]:
    return "Away_win"
  else:
    return "Draw"
df["result"] = df.apply(get_result, axis=1)
df.tail()
```

# **Creating Basic Predictors**

```python
Predictors = [
    "home_team",
    "away_team",
    "tournament",
    "country",
    "neutral",
    "year",
    "month"
]
```

# **Spliting Train & Test Split**

```python
train = df[df["year"] < 2023].copy()
test = df[df["year"] >= 2023].copy()
```

# **Encoding Categorical Variable**

```python
from sklearn.preprocessing import OrdinalEncoder

Categorical_Cols = [
    "home_team",
    "away_team",
    "tournament",
    "country"
]

encoder = OrdinalEncoder(
    handle_unknown="use_encoded_value",
    unknown_value=-1
)

train[Categorical_Cols] = encoder.fit_transform(train[Categorical_Cols])
test[Categorical_Cols] = encoder.transform(test[Categorical_Cols])
train.tail()
test.tail()
```

# **Random Forest Model**

```python
X_train = train[Predictors]
y_train = train["result"]

X_test = test[Predictors]
y_test = test["result"]

print(y_train)
```

```python
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier(
    n_estimators=750,
    max_depth=15,
    min_samples_leaf=2,
    random_state=42
)

model.fit(X_train, y_train)
```

```python
from sklearn.metrics import accuracy_score

predictions = model.predict(X_test)

accuracy = accuracy_score(y_test, predictions)
print(accuracy*100)
```

# **Feature Engineering**

Creating recent form feature for each team

1.   Calculates the points from last 5 matches
2.   Win = 3, Lose = 0 & Draw = 0
3.   Also Another feature of Recent Avg point also been created



```python
from collections import defaultdict

team_points = defaultdict(list)

home_recent_points = []
away_recent_points = []

WINDOW = 5

for _, row in df.iterrows():

    home = row["home_team"]
    away = row["away_team"]

    home_history = team_points[home][-WINDOW:]
    away_history = team_points[away][-WINDOW:]

    home_recent_points.append(sum(home_history))
    away_recent_points.append(sum(away_history))

    if row["home_score"] > row["away_score"]:

        team_points[home].append(3)
        team_points[away].append(0)

    elif row["home_score"] < row["away_score"]:

        team_points[home].append(0)
        team_points[away].append(3)

    else:

        team_points[home].append(1)
        team_points[away].append(1)


df["home_recent_points"] = home_recent_points
df["away_recent_points"] = away_recent_points
df.head()
```

```python
team_points = defaultdict(list)

avg_home_recent_points = []
avg_away_recent_points = []

WINDOW = 5

for _, row in df.iterrows():

    home = row["home_team"]
    away = row["away_team"]

    home_history = team_points[home][-WINDOW:]
    away_history = team_points[away][-WINDOW:]

    if len(home_history) == 0:
      avg_home_recent_points.append(0)
    else:
      avg_home_recent_points.append(sum(home_history) / len(home_history))

    if len(away_history) == 0:
      avg_away_recent_points.append(0)
    else:
      avg_away_recent_points.append(sum(away_history) / len(away_history))


    if row["home_score"] > row["away_score"]:

        team_points[home].append(3)
        team_points[away].append(0)

    elif row["home_score"] < row["away_score"]:

        team_points[home].append(0)
        team_points[away].append(3)

    else:

        team_points[home].append(1)
        team_points[away].append(1)


df["avg_home_recent_points"] = avg_home_recent_points
df["avg_away_recent_points"] = avg_away_recent_points
df.head()
```

Creating Head to Head statistics Features

```python

head_to_head = defaultdict(
    lambda: {
        "team1_wins": 0,
        "team2_wins": 0,
        "draws": 0
    }
)

home_h2h_wins = []
away_h2h_wins = []
h2h_draws = []

for _, row in df.iterrows():

    home = row["home_team"]
    away = row["away_team"]

    pair = tuple(sorted([home, away]))

    stats = head_to_head[pair]

    # Which team is team1?
    if pair[0] == home:

        home_h2h_wins.append(stats["team1_wins"])
        away_h2h_wins.append(stats["team2_wins"])

    else:

        home_h2h_wins.append(stats["team2_wins"])
        away_h2h_wins.append(stats["team1_wins"])

    h2h_draws.append(stats["draws"])

    if row["home_score"] > row["away_score"]:

        if pair[0] == home:
            stats["team1_wins"] += 1
        else:
            stats["team2_wins"] += 1

    elif row["home_score"] < row["away_score"]:

        if pair[0] == away:
            stats["team1_wins"] += 1
        else:
            stats["team2_wins"] += 1

    else:

        stats["draws"] += 1


df["home_h2h_wins"] = home_h2h_wins
df["away_h2h_wins"] = away_h2h_wins
df["h2h_draws"] = h2h_draws
df.tail()
```

```python
from collections import defaultdict

df = df.sort_values("date").reset_index(drop=True)

head_to_head = defaultdict(
    lambda: {
        "team1_goals": 0,
        "team2_goals": 0,
        "team1_wins": 0,
        "team2_wins": 0,
        "draws": 0,
        "matches": 0,
        "last_result": None
    }
)

home_h2h_goals_scored = []
away_h2h_goals_scored = []

home_h2h_goal_difference = []

last_h2h_result = []

h2h_matches_played = []

for _, row in df.iterrows():

    home = row["home_team"]
    away = row["away_team"]

    home_score = row["home_score"]
    away_score = row["away_score"]

    pair = tuple(sorted([home, away]))

    stats = head_to_head[pair]


    if pair[0] == home:

        home_goals = stats["team1_goals"]
        away_goals = stats["team2_goals"]

    else:

        home_goals = stats["team2_goals"]
        away_goals = stats["team1_goals"]

    home_h2h_goals_scored.append(home_goals)

    away_h2h_goals_scored.append(away_goals)

    home_h2h_goal_difference.append(home_goals - away_goals)

    h2h_matches_played.append(stats["matches"])


    if stats["last_result"] is None:

        last_h2h_result.append("No History")

    else:

        if pair[0] == home:

            last_h2h_result.append(stats["last_result"])

        else:

            if stats["last_result"] == "Win":
                last_h2h_result.append("Loss")

            elif stats["last_result"] == "Loss":
                last_h2h_result.append("Win")

            else:
                last_h2h_result.append("Draw")


    if pair[0] == home:

        stats["team1_goals"] += home_score
        stats["team2_goals"] += away_score

    else:

        stats["team1_goals"] += away_score
        stats["team2_goals"] += home_score

    stats["matches"] += 1

    if home_score > away_score:

        if pair[0] == home:

            stats["team1_wins"] += 1
            stats["last_result"] = "Win"

        else:

            stats["team2_wins"] += 1
            stats["last_result"] = "Loss"

    elif home_score < away_score:

        if pair[0] == home:

            stats["team2_wins"] += 1
            stats["last_result"] = "Loss"

        else:

            stats["team1_wins"] += 1
            stats["last_result"] = "Win"

    else:

        stats["draws"] += 1
        stats["last_result"] = "Draw"


df["home_h2h_goals_scored"] = home_h2h_goals_scored
df["away_h2h_goals_scored"] = away_h2h_goals_scored

df["home_h2h_goal_difference"] = home_h2h_goal_difference

df["last_h2h_result"] = last_h2h_result

df["h2h_matches_played"] = h2h_matches_played
df.tail()
```

```python
import numpy as np

df["home_h2h_avg_goals"] = np.where(
    df["h2h_matches_played"] == 0,
    0,
    df["home_h2h_goals_scored"] / df["h2h_matches_played"]
)

df["away_h2h_avg_goals"] = np.where(
    df["h2h_matches_played"] == 0,
    0,
    df["away_h2h_goals_scored"] / df["h2h_matches_played"]
)

df.tail()
```

Merging Current dataset with the elo rating dataset

```python
df["date"] = pd.to_datetime(df["date"])
elo_df["date"] = pd.to_datetime(
    elo_df["date"],
    format="mixed"
)
```

```python
print(elo_df["date"].dtype)
elo_df.head()
```

```python
df = df.sort_values("date")
elo_df = elo_df.sort_values("date")
```

```python
home_elo = elo_df.rename(columns={
    "team": "home_team",
    "rating": "home_elo"
})
```

```python
df = pd.merge_asof(
    df,
    home_elo[["date", "home_team", "home_elo"]],
    left_on="date",
    right_on="date",
    by="home_team",
    direction="backward",
    allow_exact_matches=False
)
```

```python
away_elo = elo_df.rename(columns={
    "team": "away_team",
    "rating": "away_elo"
})

df = pd.merge_asof(
    df,
    away_elo[["date", "away_team", "away_elo"]],
    left_on="date",
    right_on="date",
    by="away_team",
    direction="backward",
    allow_exact_matches=False
)
```

```python
df["elo_difference"] = (
    df["home_elo"] - df["away_elo"]
)
```

```python
df["home_higher_elo"] = (
    df["home_elo"] > df["away_elo"]
).astype(int)
```

```python
df["elo_ratio"] = (
    df["home_elo"] / df["away_elo"]
)
```
